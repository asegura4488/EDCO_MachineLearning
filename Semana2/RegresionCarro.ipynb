```python
datos = pd.read_csv('Datos/Auto_MPG_UCI.csv')
datos
```

```python
# Eliminar los datos faltantes
datos = datos.dropna( subset=['weight','mpg'] )
```

```python
# Separar en training y testing
```

```python
X_train, X_test, y_train, y_test = train_test_split( datos[['weight']], datos['mpg'], test_size=0.2, random_state=42 )
```

```python
# Crear el modelo
modelo = LinearRegression()
# Que pasa aca?
modelo.fit( X_train, y_train )
```

```python
# Hacer predicciones
y_pred = modelo.predict(X_test)
```

```python
# Modelo
modelo.intercept_
```

```python
modelo.coef_[0]
```

```python
# Evaluemos metricas
mean_absolute_error( y_test, y_pred )
```

```python
root_mean_squared_error( y_test, y_pred )
```

```python
r2_score(y_test, y_pred)
```

```python
plt.scatter(datos['weight'], datos['mpg'])
plt.plot(datos['weight'], modelo.intercept_ + modelo.coef_[0]*datos['weight'],color='red')
```

```python
correlacion = datos['weight'].corr(datos['mpg'])
correlacion
```

```python
predicciones = modelo.predict(X_test)
residuos = y_test - predicciones
```

```python
plt.plot(predicciones,residuos,'o')
plt.axhline(y=0,color='red')
plt.xlabel('Predicciones')
plt.ylabel('Residuos')
plt.title('Gráfico de dispersión de predicciones vs residuos')
plt.show()
```
